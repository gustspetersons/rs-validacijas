## Kas ir validācija?

Validācija ir biļetes reģistrācija konkrētā transportlīdzeklī. Iekāpjot Rīgas satiksmes sabiedriskajā transportlīdzeklī, pasažieris vai nu pieliek e-talonu pie validatora, vai skenē kvadrātkodu / ievada transportlīdzekļa identifikatoru biļeti apliecinošā lietotnē (šobrīd Rīgas satiksmes un Mobilly lietotnes).
No tā izriet, ko no datiem var un ko nevar secināt:

- Validāciju skaits ir ar e-taloniem un lietotnēm reģistrēto biļešu skaits transportlīdzekļos, nevis faktiskais cilvēku vai braucienu* skaits:
    - Pārsēžoties citā Rīgas satiksmes transportlīdzeklī un reģistrējot biļeti, rodas jauna validācija.
    - Var gadīties, ka pasažieris neveic validāciju vispār.
    - Pirmsskolas vecuma bērniem un formas tērpā esošiem dienestu darbiniekiem pienākumu pildīšanas laikā nav jāveic validācija.
    - Pasažieri ar 100 % braukšanas maksas atvieglojumu saistībā ar invaliditāti (izņemot III grupu), kuri nelieto personalizēto e-talonu, uzrādot invaliditātes apliecību, var pieprasīt vadītājam izdrukātu brīvbiļeti (Rīgas saistošo noteikumu terminoloģijā „brauciena apliecinājumu”) — biļetes veidu, kas šajos datos neparādās.
- Validācija ir biļetes reģistrācija, ne maksājums: biļete jāreģistrē arī pasažieriem ar 100 % braukšanas maksas atvieglojumiem, tāpēc arī pasažieri ar bezmaksas biļeti datos ir, bet ne visi (skatīt iepriekšējo punktu). **Sadalījums starp reģistrētajām maksas un bezmaksas biļetēm nav datu tvērumā**. Tāpat datu tvērumā nav maksas biļetes veids: 90 minūšu, mēneša biļete vai cits.
- Biļetes reģistrācijas laiks nav „transportlīdzeklī iekāpšanas laiks”: pasažieris var to reģistrēt citā laikā vai, lietotņu gadījumā, pat neatrodoties transportlīdzeklī. 
- Izkāpšanas brīdis un vieta netiek fiksēti.
- Kad visiem pasažieriem ir noteikta braukšana bez maksas, biļetes reģistrācija tiek atspējota un validāciju skaitam datos vajadzētu būt nullei.

\**Brauciens ir viena pilnīga pārvietošanās no personas sākotnējās atrašanās vietas līdz galamērķim neatkarīgi no izmantoto transportlīdzekļu skaita.*

## Datu sagatavošana

Publicētie `.txt` un `.csv` faili ir pietiekami lieli, lai būtu neērti tos katrā sesijā ielasīt no jauna, tāpēc abu kanālu dati ir saglabāti vienā DuckDB failā kā atsevišķas tabulas: *etaloni* un *lietotnes*. DuckDB failu izveido skripts `izveidot_datubazi.py` no data.gov.lv lejupielādētajiem jūnija datiem. Python vidi ar visām vajadzīgajām bibliotēkām izveido komanda `conda env create -f environment.yml`; precīzas pakotņu versijas Windows datoram ir failā `explicit.txt`. Iegulta datubāzes vadības sistēma (šajā darbā DuckDB) dod iespēju nepastarpināti uzdot SQL vaicājumus abām tabulām. Pārējais darbs ir ar pandas.

Ielasītas `DataFrame` tabulā, e-talonu validācijas aizņēma 2 450 MB. Pēc sešu kolonnu ar atkārtotām vērtībām (parks, transportlīdzekļa veids, transportlīdzekļa identifikators, maršruta nosaukums, maršruta kods un virziens) pārveidošanas par `category` tipu apjoms saruka līdz aptuveni 213 MB. Laika zīmogi, ieraksta identifikators datubāzē (`Ier_ID`) un e-talonu identifikatori (`ValidTalonaId`) ir atstāti sākotnējos tipos.

`tirit()` sakārto maršrutu galapunktu nosaukumus: avotā ir dubulti atdalītāji un dubultas atstarpes, jauktas domuzīmes ar defisēm un piecas pareizrakstības kļūdas. Šī funkcija un citas kopīgās definīcijas (vietu nosaukumu labojumi, dienu tipi, kalendāra komentāri un Jāņu ietekmes sākuma datums) ir failā `definicijas.py`. Tās izmanto arī datu sagatavošana Power BI modelim (`sagatavot_powerbi_datus.py`).

Maršrutus apzīmē maršruta kods: transportlīdzekļa veida saīsinājums un maršruta numurs, piem., A 4 ir 4. autobuss.

- **A** — autobuss;
- **Tr** — trolejbuss (nevis tramvajs);
- **Tm** — tramvajs.

Lietotņu datos nav maršruta koda, bet ir maršruta numurs un transportlīdzekļa veids, tāpēc maršruta kodu iegūst, tos apvienojot. Lietotņu datos nav arī maršruta nosaukuma, tāpēc tas ņemts no e-talonu datiem, kas ir iespējams, jo katrs lietotņu datu maršruta kods ir arī e-talonu datos.

```python
from pathlib import Path

import duckdb
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import pandas as pd

from definicijas import (
    DIENU_NOSAUKUMI,
    JANU_IETEKMES_SAKUMS,
    VIETU_LABOJUMI,
    dienas_tips,
    komentars,
    stundas_intervals,
    tirit,
    vietu_saraksts,
)

pd.set_option("styler.format.thousands", " ")
pd.set_option("styler.format.decimal", ",")
pd.set_option("styler.format.precision", 3)


# tūkstošu atdalītājs kā atstarpe; decimāldaļu atdalītājs kā komats
def fmt(n, dec=0):
    return f"{n:,.{dec}f}".replace(",", " ").replace(".", ",").replace("-", "−")


# Styler.format() neņem vērā pd.set_option atdalītājus, tāpēc skaitļu formāts
# tos saņem šeit; funkcijas formatētājam (piem., procenti) lieto parasto .format()
def skaitlu_formats(styler, **kw):
    return styler.format(decimal=",", thousands=" ", **kw)


# īpatsvars (0,213) → „21,3 %”
def procenti(x, dec=1):
    return f"{fmt(x * 100, dec)} %"


# tabulu virsraksti treknrakstā
def tabulas_virsraksts(styler, virsraksts):
    return styler.set_caption(virsraksts).set_table_styles(
        [{"selector": "caption", "props": [("font-weight", "bold")]}]
    )


# validāciju skaits y asī vizualizācijās atspoguļots ar tūkstošu atdalītāju
def fmt_yaxis(ax):
    ax.yaxis.set_major_formatter(lambda y, _: fmt(y))


# stundu ass formāts + y ass tūkstošu atdalītājs
def asu_formats(ax, step=2):
    ax.set_xlabel("Stunda")
    ax.set_xticks(range(0, 24, step))
    ax.set_xlim(0, 23)
    ax.set_ylim(bottom=0)
    ax.grid(axis="y", alpha=0.3)
    fmt_yaxis(ax)


# eksportētie attēli vienā mapē ar vienādu izšķirtspēju
FIG_DIR = Path("atteli")
FIG_DIR.mkdir(exist_ok=True)


def save_fig(fig, name, dpi=300):
    fig.savefig(FIG_DIR / f"{name}.png", dpi=dpi)


KANALU_NOSAUKUMI = {"etaloni": "E-taloni", "lietotnes": "Lietotnes"}
ETALONU_KRASA = "tab:blue"
LIETOTNU_KRASA = "tab:orange"
DARBDIENAS_KRASA = "teal"
NOGALES_KRASA = "#de6e6e"  # NOGALE = sestdiena un svētdiena

# kad sestdiena un svētdiena atspoguļotas atsevišķi:
SESTDIENAS_KRASA = "lightcoral"
SVETDIENAS_KRASA = "#cd2626"
```

```python
# abas tabulas ielasa uzreiz; savienojums aizveras bloka beigās
with duckdb.connect("RS_validacijas_2026-06.duckdb", read_only=True) as con:
    et = con.sql("SELECT * FROM etaloni").df()
    liet = con.sql("SELECT * FROM lietotnes").df()


def dienas_pievienosana(kanals, datuma_kolonna):
    kanals["diena"] = pd.Categorical(
        kanals[datuma_kolonna].dt.day_name().map(DIENU_NOSAUKUMI),
        categories=list(DIENU_NOSAUKUMI.values()),
        ordered=True,
    )
    return kanals


# --- etaloni: katra validācija savā rindā ---------------------------------------------
ier_id_ir_unikals = et["Ier_ID"].is_unique
assert ier_id_ir_unikals, "Ier_ID nav unikāls — vienai validācijai jābūt vienai rindai"

mem_before = et.memory_usage(deep=True).sum() / 1e6
for col in ["Parks", "TranspVeids", "GarNr", "MarsrNos", "TMarsruts", "Virziens"]:
    et[col] = et[col].astype("category")
mem_after = et.memory_usage(deep=True).sum() / 1e6

tiri_nosaukumi = {c: tirit(c) for c in et["MarsrNos"].cat.categories}
et["marsr_nos_tirs"] = et["MarsrNos"].map(tiri_nosaukumi).astype("category")

# katram labojumam avotā jābūt savai vietai, citādi tas neko nelabo
avota_vietas = {v for n in et["MarsrNos"].cat.categories for v in vietu_saraksts(n)}
neatrastie_labojumi = VIETU_LABOJUMI.keys() - avota_vietas
assert not neatrastie_labojumi, f"labojumi bez vietas avotā: {neatrastie_labojumi}"

et["datums"] = et["Laiks"].dt.normalize()
et = dienas_pievienosana(et, "datums")
et["stunda"] = et["Laiks"].dt.hour
marsrutu_numuri = {c: c.split()[-1] for c in et["TMarsruts"].cat.categories}
et["marsr_nr"] = et["TMarsruts"].map(marsrutu_numuri).astype("category")

# --- lietotnes: validāciju skaits oriģinālajā failā apkopots kolonnā Skaits -----------
for col in ["Stunda", "Transports", "Marsruts"]:
    liet[col] = liet[col].astype("category")

# „16:00-17:00” → „16”
liet["stunda"] = liet["Stunda"].str[:2].astype(int)
liet = dienas_pievienosana(liet, "Datums")

print(
    f"Tabula ar validācijām no e-taloniem:"
    f"{fmt(len(et)):>10} rindas → {fmt(len(et)):>10} validācijas"
)
print(
    f"Tabula ar validācijām no lietotnēm: "
    f"{fmt(len(liet)):>10} rindas → {fmt(liet['Skaits'].sum()):>10} validācijas"
)
print(
    f"E-talonu tabula atmiņā pirms un pēc kolonnu pārveidošanas par category: "
    f"{fmt(mem_before)} → {fmt(mem_after)} MB"
)
```

```python
JANI = pd.to_datetime(["2026-06-23", "2026-06-24"])

kalendars = pd.DataFrame({"datums": pd.date_range("2026-06-01", "2026-06-30")})
kalendars = dienas_pievienosana(kalendars, "datums")

kalendars["dienas_tips"] = kalendars["datums"].map(dienas_tips)
kalendars["komentari"] = kalendars["datums"].map(komentars)
kalendars = kalendars.set_index("datums")
et["dienas_tips"] = et["datums"].map(kalendars["dienas_tips"])
liet["dienas_tips"] = liet["Datums"].map(kalendars["dienas_tips"])

# katrs datums ir kalendārā → neviens tips nav palicis tukšs
assert et["dienas_tips"].notna().all() and liet["dienas_tips"].notna().all()

kalendars["validacijas_no_etaloniem"] = et.resample("D", on="Laiks").size()

# dienām bez rindām resample summa dod 0, nevis NaN — 23.–24. jūnijs paliek 0
kalendars["validacijas_no_lietotnem"] = liet.resample("D", on="Datums")["Skaits"].sum()
kalendars["validacijas_kopa"] = (
    kalendars["validacijas_no_etaloniem"] + kalendars["validacijas_no_lietotnem"]
)
kalendars["lietotnu_validaciju_ipatsvars"] = (
    kalendars["validacijas_no_lietotnem"] / kalendars["validacijas_kopa"]
)

# dienu summas sakrīt ar ielasīto → apkopojumā nekas nav pazudis
assert kalendars["validacijas_no_etaloniem"].sum() == len(et)
assert kalendars["validacijas_no_lietotnem"].sum() == liet["Skaits"].sum()

KALENDARA_KOLONNAS = {
    "diena": "Diena",
    "dienas_tips": "Dienas tips",
    "validacijas_no_etaloniem": "No e-taloniem",
    "validacijas_no_lietotnem": "No lietotnēm",
    "validacijas_kopa": "Validācijas kopā",
    "lietotnu_validaciju_ipatsvars": "Lietotņu validāciju īpatsvars",
    "komentari": "Komentārs",
}

ar_komentariem = kalendars["komentari"].ne("").sum()
print(f"kalendārs: {len(kalendars)} dienas, {ar_komentariem} ar komentāriem")
```

Atskaitot iepriekšminētos 10 maršrutus, lietotņu validāciju īpatsvara diapazons pa maršrutiem 1.–19. jūnijā ir **9–33 %**.

```python
# viena rinda katram (transporta veids, numurs) pārim, 1.–19. jūnijā, abi kanāli blakus
marsr_et = (
    et_pj.groupby(["TranspVeids", "marsr_nr", "TMarsruts"], observed=True)
    .size()
    .reset_index(name="et_skaits")
)
assert not marsr_et.duplicated(["TranspVeids", "marsr_nr"]).any()

marsr_liet = (
    liet_pj.groupby(["Transports", "Marsruts"], observed=True)["Skaits"]
    .sum()
    .reset_index()
    .rename(
        columns={
            "Transports": "TranspVeids",
            "Marsruts": "marsr_nr",
            "Skaits": "liet_skaits",
        }
    )
)
marsruti = pd.merge(
    marsr_et,
    marsr_liet,
    on=["TranspVeids", "marsr_nr"],
    how="outer",
    validate="one_to_one",
)

# agrākais nosaukums, ko maršruts jūnijā „nēsāja”
pirmas_rindas = et.groupby(["TranspVeids", "marsr_nr"], observed=True)["Laiks"].idxmin()
senakie_nosaukumi = et.loc[pirmas_rindas, ["TranspVeids", "marsr_nr", "marsr_nos_tirs"]]
senakie_nosaukumi = senakie_nosaukumi.rename(columns={"marsr_nos_tirs": "nosaukums"})
marsruti = pd.merge(
    marsruti,
    senakie_nosaukumi,
    on=["TranspVeids", "marsr_nr"],
    how="left",
    validate="one_to_one",
)

# lietotņu datos trūkstošajiem maršrutiem validāciju skaits aizpildīts ar 0
# kā vesels skaitlis
marsruti["liet_skaits"] = marsruti["liet_skaits"].fillna(0).astype("int64")
marsruti["kopa"] = marsruti["et_skaits"] + marsruti["liet_skaits"]
marsruti["liet_iptsv"] = marsruti["liet_skaits"] / marsruti["kopa"]

# apakšsummas sakrīt ar avota tabulām
assert marsruti["et_skaits"].sum() == len(et_pj)
assert marsruti["liet_skaits"].sum() == liet_pj["Skaits"].sum()

liet_iptsv_tikla_pj = marsruti["liet_skaits"].sum() / marsruti["kopa"].sum()

# atlase: bez nakts maršrutiem, A 71 (vienreizējs) un A 35 (slēgts); sk. B pielikumu
regularie_marsruti = marsruti[
    ~marsruti["marsr_nr"].astype(str).str.startswith("N")
    & ~marsruti["TMarsruts"].isin(["A 35", "A 71"])
]
augsa = regularie_marsruti["liet_iptsv"].max()
apaksa = regularie_marsruti["liet_iptsv"].min()

# tie paši maršruti visā jūnijā (ieskaitot Jāņus) — diapazons salīdzinājumam
et_jun = et.groupby(["TranspVeids", "marsr_nr"], observed=True).size()
liet_jun = (
    liet.rename(columns={"Transports": "TranspVeids", "Marsruts": "marsr_nr"})
    .groupby(["TranspVeids", "marsr_nr"], observed=True)["Skaits"]
    .sum()
)
kopa_jun = et_jun.add(liet_jun, fill_value=0)
regularie_atslegas = pd.MultiIndex.from_frame(
    regularie_marsruti[["TranspVeids", "marsr_nr"]]
)
liet_iptsv_jun = (liet_jun / kopa_jun).reindex(regularie_atslegas)
liet_iptsv_tikla_jun = liet["Skaits"].sum() / (len(et) + liet["Skaits"].sum())

print(
    f"Lietotņu validāciju īpatsvars tīklā: "
    f"viss jūnijs {fmt(liet_iptsv_tikla_jun * 100, 1)} %; "
    f"1.–19. jūnijs {fmt(liet_iptsv_tikla_pj * 100, 1)} %."
)
print(
    f"{len(marsruti)} maršruti 1.–19. jūnijā, "
    f"no tiem {len(regularie_marsruti)} regulārie dienas maršruti."
)
print(
    f"Lietotņu īpatsvars regulārajos dienas maršrutos 1.–19. jūnijā: "
    f"no {fmt(apaksa * 100, 1)} % līdz {fmt(augsa * 100, 1)} % "
    f"(lielākais {fmt(augsa / apaksa, 1)} reizes pārsniedz mazāko)."
)
print(
    f"Tiem pašiem maršrutiem visā jūnijā: "
    f"no {fmt(liet_iptsv_jun.min() * 100, 1)} % "
    f"līdz {fmt(liet_iptsv_jun.max() * 100, 1)} %."
)
```

```python
def marsrutu_attiecibas(diena_a, diena_b):
    # e-talonu validāciju attiecība starp abām dienām, visiem maršrutiem uzreiz;
    # maršruts bez validācijām vienā no dienām paliek bez attiecības (NaN)
    izlase = et[et["datums"].isin(pd.to_datetime([diena_a, diena_b]))]
    skaiti = (
        izlase.groupby(["TMarsruts", "datums"], observed=True).size().unstack("datums")
    )
    attieciba = skaiti[pd.Timestamp(diena_a)] / skaiti[pd.Timestamp(diena_b)]
    return attieciba.round(2).sort_values(ascending=False)


# svētdienu pāris: gadatirgus un saule (7.) pret mākoņainu un vēsu svētdienu (14.)
svetdienas = marsrutu_attiecibas("2026-06-07", "2026-06-14")

# sestdienu pāris: gadatirgus (6.) pret 13. jūniju; laikapstākļi abās mainīgi
sestdienas = marsrutu_attiecibas("2026-06-06", "2026-06-13")

nosaukumi = marsruti.set_index("TMarsruts")["nosaukums"]

# fmt: off
MARSRUTU_KOMENTARI = {
    "Tr 31":  "gar Brīvdabas muzeju; gadatirgum pastiprināts",
    "A 28":   "Juglas galapunkts pie muzeja; gadatirgum pastiprināts",
    "A 58":   "vasarā daļa reisu līdz Vecāķu pludmalei",
    "A 36":   "beidzas pie pludmales; gadatirgus ne skar, ne tam pastiprina",
    "A 48":   "galapunkts pie Pļavnieku kapiem — kapu apmeklējumi labos laikapstākļos",
    "A 29":   "pludmale (Vecāķi), bet gadatirgum pastiprināts",
    "A 54":   "galapunkts pie Botāniskā dārza — āra galamērķis",
    "A 24":   "Mangaļsala — pludmale un mols",
    "Tm 11":  "Mežaparks — atpūta; uz gadatirgu tieši neved",
    "A 39":   "galapunkts pie Lāčupes kapiem — kapu apmeklējumi labos laikapstākļos",
    "A 56":   "Daugavgrīvas gals pie pludmales",
    "A 18":   "galapunkts Dārziņos — mazdārziņu apkaime pie Daugavas",
    "Tm 1":   "pastiprināts gadatirgum, tomēr uzvedās kā kontrole. Ved cauri visai "
              "Rīgai, un galapunkts „Jugla” ir netālu no gadatirgus",
    "A 22":   "kontrole — lidostas pieprasījumu nosaka lidojumu grafiks",
}
# fmt: on

attiecibas = pd.DataFrame({"svetdiena": svetdienas, "sestdiena": sestdienas})

# tāpat kā maršrutu salīdzinājumā: bez nakts maršrutiem, A 71 un A 35; sk. B pielikumu
attiecibas = attiecibas[attiecibas.index.isin(regularie_marsruti["TMarsruts"])]
attiecibas.insert(0, "nosaukums", nosaukumi)
attiecibas["komentars"] = (
    pd.Series(MARSRUTU_KOMENTARI).reindex(attiecibas.index).fillna("")
)
ATTIECIBU_KOLONNAS = {
    "nosaukums": "Maršruta nosaukums",
    "svetdiena": "Svētdiena 07./14.",
    "sestdiena": "Sestdiena 06./13.",
    "komentars": "Komentārs",
}


def attiecibu_tabula(df):
    return (
        df.rename(columns=ATTIECIBU_KOLONNAS)
        .rename_axis("Maršruts")
        .style.pipe(skaitlu_formats, precision=2)
    )


tikla_attieciba = round(
    (et["datums"] == "2026-06-07").sum() / (et["datums"] == "2026-06-14").sum(), 2
)
print(
    "Sakārtots dilstoši pēc svētdienu (07./14.) attiecības; tīkla kopējā"
    f" attiecība šajās svētdienās ir {fmt(tikla_attieciba, 2)}."
)
attiecibu_tabula(attiecibas.nlargest(15, "svetdiena"))
```

Lietotņu validācijas tajā pašā laikā ir +1,5 % un tad −0,8 % — kopumā pat ar nelielu pieaugumu (+0,6 %). Iespējams, to veicina arī tūristi, kas varētu būt vairāk orientēti uz lietotnēm.

Pareizais secinājums tātad ir: **jūnija pirmajās trīs nedēļās e-talonu validācijas lēni sarūk par 2–3 % nedēļā, kamēr lietotņu validācijas saglabā līmeni**. Sākotnēji redzamais kritums ir divu neparastu dienu ietekme, nevis tendences lūzums. 

```python
def nedelu_raditaji(validacijas):
    # e-talonu validācijas, unikālie e-taloni un validācijas uz e-talonu pa nedēļām
    ned = validacijas.resample("W", on="Laiks").agg(
        etalonu_validacijas=("ValidTalonaId", "size"),
        unikalie_etaloni=("ValidTalonaId", "nunique"),
    )
    ned["uz_etalonu"] = ned["etalonu_validacijas"] / ned["unikalie_etaloni"]
    return ned.rename_axis("Nedēļa līdz")


ned = nedelu_raditaji(et[et["datums"] <= "2026-06-21"])
NED_KOLONNAS = {
    "etalonu_validacijas": "E-talonu validācijas",
    "unikalie_etaloni": "Unikālie e-taloni",
    "uz_etalonu": "Validācijas uz e-talonu",
}
ned.rename(columns=NED_KOLONNAS).style.format_index("{:%Y-%m-%d}")
```

*`ValidTalonaId` skaita **e-talonus, nevis cilvēkus**: vienam cilvēkam var būt vairāki e-taloni, un vienu e-talonu var lietot vairāki cilvēki.*

*`uz_etalonu` ir **nedēļas** rādītājs — vidēji ~7 validācijas uz e-talonu nedēļā jeb aptuveni viena dienā.*

```python
IPATNEJIE = ["A 35", "A 71", "A 95"]

# cik transportlīdzekļu maršrutā vidēji vienā dienā
transport_skaits_diena = (
    et[et["TMarsruts"].isin(IPATNEJIE)]
    .groupby(["TMarsruts", "datums"], observed=True)["GarNr"]
    .nunique()
    .groupby("TMarsruts", observed=True)
    .mean()
    .round(1)
)

ipatnejie = (
    et[et["TMarsruts"].isin(IPATNEJIE)]
    .groupby("TMarsruts", observed=True)
    .agg(
        validacijas_kopa=("Ier_ID", "size"),
        dienas=("datums", "nunique"),
        pirma=("datums", "min"),
        pedeja=("datums", "max"),
        transport_skaits=("GarNr", "nunique"),
        parki=("Parks", lambda s: ", ".join(sorted(s.astype(str).unique()))),
    )
)
ipatnejie["transport_skaits_diena"] = transport_skaits_diena

# visi trīs īpatnējie ir autobusi; lietotņu failā maršruts = (Transports, numurs)
ipatnejie["lietotnu_skaits"] = [
    int(
        liet.loc[
            (liet["Transports"] == "Autobuss") & (liet["Marsruts"] == kods.split()[-1]),
            "Skaits",
        ].sum()
    )
    for kods in ipatnejie.index
]
KOLONNU_NOSAUKUMI = {
    "parki": "Transportlīdzekļu parki",
    "transport_skaits": "Transportlīdzekļu skaits",
    "dienas": "Kursēšanas dienu skaits jūnijā",
    "transport_skaits_diena": "Vidējais transportlīdzekļu skaits dienā",
    "pirma": "Pirmā kursēšanas diena",
    "pedeja": "Pēdējā kursēšanas diena",
    "validacijas_kopa": "Validācijas no e-taloniem",
    "lietotnu_skaits": "Validācijas no lietotnēm",
}
ipatnejie = (
    ipatnejie[list(KOLONNU_NOSAUKUMI)]
    .rename(columns=KOLONNU_NOSAUKUMI)
    .rename_axis("Maršruta kods")
)
ipatnejie.style.pipe(skaitlu_formats, precision=1).format(
    "{:%Y-%m-%d}", subset=["Pirmā kursēšanas diena", "Pēdējā kursēšanas diena"]
)
```

```python
# autobusu parku īpatsvars jūnija autobusu validācijās un autobusu skaitā
autobusu_parki = (
    et[et["TranspVeids"] == "Autobuss"]
    .groupby("Parks", observed=True)
    .agg(validacijas=("Ier_ID", "size"), autobusi=("GarNr", "nunique"))
)
(
    (autobusu_parki / autobusu_parki.sum())
    .rename(
        columns={
            "validacijas": "Īpatsvars autobusu validācijās",
            "autobusi": "Īpatsvars autobusu skaitā",
        }
    )
    .style.format(procenti)
    .pipe(tabulas_virsraksts, "Autobusu parki jūnijā")
)
```

## Paplašināšanas iespējas

Pieejamākais datu analīzes paplašinājums ir **e-talonu validācijām**: Rīgas satiksmes trešā data.gov.lv publicētā datu kopa ir **Maršrutu saraksti** (GTFS) ar pieturām, maršrutu ģeometriju un kustības sarakstiem. Tā ļautu:

- **sagrupēt maršrutus pēc šķērsotajām apkaimēm** — pārbaudīt, kā lietotņu īpatsvars atšķiras pa apkaimēm un vai to ietekmē centra šķērsošana, ko maršruta nosaukums neatklāj (piem., 1. tramvajs Imanta–Jugla).
- **novērtēt pieprasījumu pieturu līmenī** — e-talonu validācijām ir laiks, transportlīdzeklis, maršruts un virziens. Salāgojot tos ar kustības sarakstu, var novērtēt iekāpšanas pieturu (ar piebildi, ka validācijas brīdis var atšķirties no faktiskās iekāpšanas, tāpēc pietura nosakāma aptuveni un kopsummās kalibrējama). Tādējādi var novērtēt pieprasījumu katrā pieturā, ne tikai visā maršrutā kopā.
- **atdalīt pieprasījumu no piedāvājuma** — validācijas uz vienu plānoto reisu parāda, vai maršruta pieaugums ir pasažieru pieaugums vai papildu reisi; piemēram, +175 % 7. jūnija 31. trolejbusam Centrāltirgus–Berģuciems, kam tika norīkoti papildreisi Brīvdabas muzeja gadatirgus dēļ.
- **precizēt pārsēšanās ķēdes** — secīgas validācijas pēc `ValidTalonaId` laika logā var ķēdēt braucienos jau ar pašreizējiem datiem; GTFS ģeometrija ļautu precīzāk segmentēt un vizualizēt braucienus, kas laika logā izskatās pēc pārsēšanās.

**Viss telpiskais griezums attiecas tikai uz e-talonu kanālu: lietotņu dati ir apkopoti pa datumiem, stundām un maršrutiem — bez validācijas laika, transportlīdzekļa identifikatora, virziena un biļetes identifikatora.**

## C pielikums — 7. jūnija sadalījums pa maršrutiem un kanāliem

Astoņi maršruti sakārtoti pēc izmaiņu validāciju skaitā secinātā cēloņa: gadatirgus, laikapstākļi, abi cēloņi, kontrole.

```python
MARSRUTI_IZVERTESANAI = [
    "Tr 31",
    "A 28",
    "A 36",
    "A 58",
    "Tm 11",
    "A 29",
    "Tm 1",
    "A 22",
]
attiecibu_tabula(attiecibas.loc[MARSRUTI_IZVERTESANAI])
```

- **Tr 31 Centrāltirgus–Berģuciems (2,75 / 2,20)** un **A 28 Jugla–Langstiņi (1,98 / 1,52)** paliek augstu abās dienās → gadatirgus. Abi ir ikdienas maršruti uz dzīvojamām apkaimēm, ne uz atpūtas vietām, tāpēc laikapstākļu loma tajos ir maza — sestdienas +120 % un +52 % ir pasākuma ietekme.
- **A 36 Imanta–Vakarbuļļi (1,70 / 0,90)**, **A 58 Purvciems – Vecmīlgrāvis – Vecāķi (1,82 / 0,99)** un **Tm 11 Ausekļa iela – Mežaparks (1,41 / 0,78)** ir virs 1 svētdien un ap vai *zem* 1 sestdien → laikapstākļi. Rīgas satiksme nevienu no tiem gadatirgum nepastiprināja, un gadatirgus sestdienā neviens nepieauga, tātad gadatirgus tos tieši neskāra.
- **A 29 Mežciems – Vecmīlgrāvis – Vecāķi (1,57 / 1,24)** ved uz pludmali un kursē gar Brīvdabas muzeju: tas ir vienīgais pludmales maršruts virs 1 sestdien, un Rīgas satiksme to gadatirgum pastiprināja.
- **Tm 1 Imanta–Jugla (1,15 / 0,98)** gadatirgum pastiprināts, tomēr attiecībās uzvedas kā kontrole.
- **A 22 Abrenes iela – Lidosta (1,01 / 1,06)** gadatirgus neietekmēts: atrodas tālu no gadatirgus, un pasažieru skaitu galvenokārt nosaka lidojumu grafiks, tāpēc izmantojams kā kontrole.

A 29 un Tm 1 laikapstākļu ietekmes novērtēšanai neder, jo abus skāra arī gadatirgus. Tāpēc laikapstākļu ietekme uz kanālu sadalījumu vērtēta trijos maršrutos, kurus gadatirgus neskāra: A 36, A 58 un Tm 11.

```python
# abu kanālu validācijas abās svētdienās; lietotņu datiem maršruta kods
# (piem., A 36) pievienots no e-talonu maršrutu tabulas
et_svetdienas = (
    et[et["datums"].isin([GADATIRGUS_SVETDIENA, KONTROLES_SVETDIENA])]
    .groupby(["TMarsruts", "datums"], observed=True)
    .size()
    .unstack("datums")
    .reindex(MARSRUTI_IZVERTESANAI)
)
liet_svetdienas = (
    liet[liet["Datums"].isin([GADATIRGUS_SVETDIENA, KONTROLES_SVETDIENA])]
    .rename(columns={"Transports": "TranspVeids", "Marsruts": "marsr_nr"})
    .merge(
        marsruti[["TranspVeids", "marsr_nr", "TMarsruts"]],
        on=["TranspVeids", "marsr_nr"],
        how="left",
    )
    .groupby(["TMarsruts", "Datums"], observed=True)["Skaits"]
    .sum()
    .unstack("Datums")
    .reindex(MARSRUTI_IZVERTESANAI)
)
e7, e14 = et_svetdienas[GADATIRGUS_SVETDIENA], et_svetdienas[KONTROLES_SVETDIENA]
l7, l14 = liet_svetdienas[GADATIRGUS_SVETDIENA], liet_svetdienas[KONTROLES_SVETDIENA]

# īpatsvaru papildu validācijās rāda tikai maršrutos, kur abu kanālu kopsumma
# pieauga vismaz par 10 %; mazākai starpībai īpatsvars ir nestabils
auga = (e7 + l7) / (e14 + l14) >= 1.1
svetdienu_kanali = pd.DataFrame(
    {
        "E-taloni 07./14.": e7 / e14,
        "Lietotnes 07./14.": l7 / l14,
        "Lietotņu īpatsvars 07.06.": l7 / (e7 + l7),
        "Lietotņu īpatsvars 14.06.": l14 / (e14 + l14),
        "Lietotņu īpatsvars papildu validācijās": (
            (l7 - l14) / ((e7 - e14) + (l7 - l14))
        ).where(auga),
    }
).rename_axis("Maršruts")
(
    svetdienu_kanali.style.pipe(skaitlu_formats, precision=2)
    .format(
        procenti,
        subset=[c for c in svetdienu_kanali if "īpatsvars" in c],
        na_rep="—",
    )
    .pipe(
        tabulas_virsraksts,
        "Svētdienu attiecība 07./14. un lietotņu īpatsvars astoņos maršrutos"
        " (papildu validāciju īpatsvars tikai maršrutos ar pieaugumu ≥ 10 %)",
    )
)
```

Visos trijos laikapstākļu maršrutos 7. jūnijā auga abi kanāli, bet kanālu sadalījums mainījās dažādos virzienos: A 36 lietotņu īpatsvars papildu validācijās ir krietni virs 14. jūnija īpatsvara, A 58 zem tā, Tm 11 nedaudz virs. Kontroles maršrutā A 22 lietotņu īpatsvars gandrīz nemainījās.
